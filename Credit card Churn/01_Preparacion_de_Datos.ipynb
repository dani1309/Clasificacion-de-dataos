### 6. Reglas de calidad y atípicos (Tabla 4 del informe)
Al inspeccionar las variables categóricas, encontramos que `Education_Level`, `Marital_Status` e `Income_Category` tienen registros con el texto `'Unknown'`.


En cuanto a las numéricas, revisamos mínimos y máximos para comprobar que no hayan edades negativas ni porcentajes mayores a 1.


```python
# Revisión de rangos de variables numéricas
print(df_limpio.describe().T[['min', 'mean', 'max']].round(2))

# Reemplazamos los 'Unknown' por NaN para tratarlos como valores nulos
df_nulos = df_limpio.replace('Unknown', np.nan)
```

```python
# ==============================================================================
# TABLA DE REGLAS DE CALIDAD - FORMATO OFICIAL CRISP-DM (4 COLUMNAS)
# ==============================================================================
d = df_limpio

# Diccionario con: (Regla de negocio, Condición de error, Formato observado)
reglas_oficiales = {
    'Attrition_Flag': (
        "Existing Customer, Attrited Customer",
        ~d['Attrition_Flag'].isin(['Existing Customer', 'Attrited Customer']),
        "Existing Customer (83.9%), Attrited Customer (16.1%)"
    ),
    'Customer_Age': (
        "18 - 100 años",
        ~d['Customer_Age'].between(18, 100),
        f"{d['Customer_Age'].min()} - {d['Customer_Age'].max()} años"
    ),
    'Gender': (
        "M, F",
        ~d['Gender'].isin(['M', 'F']),
        "F (52.9%), M (47.1%)"
    ),
    'Dependent_count': (
        "0 - 10 personas",
        ~d['Dependent_count'].between(0, 10),
        f"{d['Dependent_count'].min()} - {d['Dependent_count'].max()} dependientes"
    ),
    'Education_Level': (
        "6 niveles educativos formales",
        ~d['Education_Level'].isin(['Uneducated', 'High School', 'College', 'Graduate', 'Post-Graduate', 'Doctorate']),
        "Incluye 'Unknown' en 1,519 registros (15.0%)"
    ),
    'Marital_Status': (
        "Married, Single, Divorced",
        ~d['Marital_Status'].isin(['Married', 'Single', 'Divorced']),
        "Incluye 'Unknown' en 749 registros (7.4%)"
    ),
    'Income_Category': (
        "5 rangos de ingreso en USD",
        ~d['Income_Category'].isin(['Less than $40K', '$40K - $60K', '$60K - $80K', '$80K - $120K', '$120K +']),
        "Incluye 'Unknown' en 1,112 registros (11.0%)"
    ),
    'Card_Category': (
        "Blue, Silver, Gold, Platinum",
        ~d['Card_Category'].isin(['Blue', 'Silver', 'Gold', 'Platinum']),
        "Blue, Silver, Gold, Platinum"
    ),
    'Months_on_book': (
        "> 0 y <= (edad - 18) x 12",
        (d['Months_on_book'] <= 0) | (d['Months_on_book'] > (d['Customer_Age'] - 18) * 12),
        f"{d['Months_on_book'].min()} - {d['Months_on_book'].max()} meses"
    ),
    'Total_Relationship_Count': (
        "Entero >= 1 (mínimo la tarjeta)",
        d['Total_Relationship_Count'] < 1,
        f"{d['Total_Relationship_Count'].min()} - {d['Total_Relationship_Count'].max()} productos"
    ),
    'Months_Inactive_12_mon': (
        "0 - 12 meses",
        ~d['Months_Inactive_12_mon'].between(0, 12),
        f"{d['Months_Inactive_12_mon'].min()} - {d['Months_Inactive_12_mon'].max()} meses"
    ),
    'Contacts_Count_12_mon': (
        "Conteo >= 0",
        d['Contacts_Count_12_mon'] < 0,
        f"{d['Contacts_Count_12_mon'].min()} - {d['Contacts_Count_12_mon'].max()} contactos"
    ),
    'Credit_Limit': (
        "> $0 USD",
        d['Credit_Limit'] <= 0,
        f"${d['Credit_Limit'].min():,.1f} - ${d['Credit_Limit'].max():,.1f}"
    ),
    'Total_Revolving_Bal': (
        "Entre $0 y Credit_Limit",
        (d['Total_Revolving_Bal'] < 0) | (d['Total_Revolving_Bal'] > d['Credit_Limit']),
        f"${d['Total_Revolving_Bal'].min():,.0f} - ${d['Total_Revolving_Bal'].max():,.0f}"
    ),
    'Avg_Open_To_Buy': (
        ">= 0 e igual a Límite - Saldo",
        (d['Avg_Open_To_Buy'] < 0) | ((d['Avg_Open_To_Buy'] - (d['Credit_Limit'] - d['Total_Revolving_Bal'])).abs() > 1),
        f"${d['Avg_Open_To_Buy'].min():,.1f} - ${d['Avg_Open_To_Buy'].max():,.1f}"
    ),
    'Total_Amt_Chng_Q4_Q1': (
        "Razón >= 0",
        d['Total_Amt_Chng_Q4_Q1'] < 0,
        f"{d['Total_Amt_Chng_Q4_Q1'].min():.3f} - {d['Total_Amt_Chng_Q4_Q1'].max():.3f}"
    ),
    'Total_Trans_Amt': (
        "Monto >= $0 USD",
        d['Total_Trans_Amt'] < 0,
        f"${d['Total_Trans_Amt'].min():,.0f} - ${d['Total_Trans_Amt'].max():,.0f}"
    ),
    'Total_Trans_Ct': (
        "Conteo >= 0",
        d['Total_Trans_Ct'] < 0,
        f"{d['Total_Trans_Ct'].min()} - {d['Total_Trans_Ct'].max()} transacciones"
    ),
    'Total_Ct_Chng_Q4_Q1': (
        "Razón >= 0",
        d['Total_Ct_Chng_Q4_Q1'] < 0,
        f"{d['Total_Ct_Chng_Q4_Q1'].min():.3f} - {d['Total_Ct_Chng_Q4_Q1'].max():.3f}"
    ),
    'Avg_Utilization_Ratio': (
        "Entre 0.0 y 1.0",
        ~d['Avg_Utilization_Ratio'].between(0, 1),
        f"{d['Avg_Utilization_Ratio'].min():.3f} - {d['Avg_Utilization_Ratio'].max():.3f}"
    ),
}

filas_tabla = []
for col, (regla, incumple, rango_obs) in reglas_oficiales.items():
    n_error = int(incumple.sum())
    filas_tabla.append({
        'Variable': col,
        'Regla de calidad': regla,
        'Mínimo-Máximo/categorías de los datos': rango_obs,
        'Indicativo de error': 'Error' if n_error > 0 else 'Sin error'
    })

tabla_calidad_oficial = pd.DataFrame(filas_tabla)
pd.set_option('display.max_colwidth', None)
tabla_calidad_oficial
```

```python
# Detección de atípicos con el criterio IQR (1.5 veces el rango intercuartílico)
num_cols = df_limpio.select_dtypes(include=[np.number]).columns
filas_iqr = []
for col in num_cols:
    q1, q3 = df_limpio[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_atip = ((df_limpio[col] < lim_inf) | (df_limpio[col] > lim_sup)).sum()
    filas_iqr.append({'Variable': col, 'Q1': q1, 'Q3': q3, 'IQR': iqr,
                      'Límite inferior': lim_inf, 'Límite superior': lim_sup,
                      'Atípicos': n_atip, '% atípicos': round(n_atip / len(df_limpio) * 100, 2)})

tabla_iqr = pd.DataFrame(filas_iqr).sort_values('% atípicos', ascending=False).reset_index(drop=True)
display(tabla_iqr[['Variable', 'Atípicos', '% atípicos']])

# Boxplots de las variables numéricas
fig, axes = plt.subplots(4, 4, figsize=(16, 14))
for ax, col in zip(axes.flat, num_cols):
    sns.boxplot(y=df_limpio[col], ax=ax, color='#1f77b4')
    ax.set_title(col, fontsize=10, fontweight='bold')
    ax.set_ylabel('')
for ax in axes.flat[len(num_cols):]:
    ax.set_visible(False)
plt.suptitle("Boxplots de variables numéricas (detección de atípicos con IQR)", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Decisión sobre reglas de calidad y atípicos**

La tabla de reglas de calidad muestra que ninguna variable numérica incumple su regla de negocio. Las edades están dentro del rango, los conteos y montos no son negativos, el saldo rotativo nunca supera el cupo y la razón de utilización está entre 0 y 1. Las únicas variables con error son Education_Level, Marital_Status e Income_Category, por la categoría Unknown, que no es un valor real sino un dato que el banco no registró, y por eso la convertimos en nulo para tratarla en la sección de nulos.

Con el criterio del IQR, las variables con más atípicos son Credit_Limit (9.72%), Avg_Open_To_Buy (9.51%), Total_Trans_Amt (8.85%) y Contacts_Count_12_mon (6.21%), mientras que Customer_Age y Total_Trans_Ct solo tienen 2 atípicos cada una. Decidimos conservar todos estos registros porque son valores reales del negocio y no errores de captura. Los cupos y montos altos corresponden a clientes con mayor capacidad de crédito o que usan mucho la tarjeta, y en los conteos de contactos, inactividad y antigüedad el IQR marca atípicos solo porque la mayoría de clientes se concentra en pocos valores. Además, eliminar filas quitaría información útil para predecir el abandono y reduciría todavía más la clase minoritaria.


### 7. Limpieza de valores nulos (Regla del 15% - Tabla 5)
Aplicamos los tres pasos que pide la plantilla del informe:
1. **Eliminar filas con más del 15% de nulos:**  
   En nuestras 19 variables predictoras, si una fila tiene 3 valores faltantes tiene $3/19 = 15.79\% > 15\%$. Se encontraron 7 filas con esta condición y las eliminamos.
2. **Eliminar columnas con más del 15% de nulos:**  
   `Education_Level` aparece con 15.00% en la tabla, pero ese valor sale del redondeo a dos decimales. En realidad tiene 1519 nulos sobre 10127 registros, es decir $1519/10127 = 14.9995\%$, que queda por debajo del 15% y por eso la columna se conserva y se imputa en vez de eliminarse (el cálculo exacto se muestra en la celda que sigue a la limpieza). `Income_Category` tiene 10.98% y `Marital_Status` tiene 7.40%. Ninguna supera el 15%, por lo que no se borra ninguna columna.
3. **Imputar variables con menos del 15% de nulos:**  
   Como las tres variables son categóricas, se imputan con la **moda** de cada una:
   - `Education_Level` -> Moda: `'Graduate'`
   - `Income_Category` -> Moda: `'Less than $40K'`
   - `Marital_Status` -> Moda: `'Married'`


```python
# 1. Ver porcentaje de nulos por columna
porc_nulos = (df_nulos.isnull().sum() / len(df_nulos) * 100).round(2)
print("Porcentaje de nulos por columna:")
print(porc_nulos[porc_nulos > 0])

# 2. Eliminar registros con más del 15% de nulos (3 de 19 variables = 15.79%)
pred_cols = [c for c in df_nulos.columns if c != 'Attrition_Flag']
nulos_por_fila = (df_nulos[pred_cols].isnull().sum(axis=1) / len(pred_cols)) * 100
filas_a_borrar = (nulos_por_fila > 15.0).sum()
print(f"\nFilas eliminadas con > 15% de nulos: {filas_a_borrar}")

df_preparado = df_nulos.loc[nulos_por_fila <= 15.0].copy()

# 3. Imputar las columnas categóricas con su moda
for col in ['Education_Level', 'Income_Category', 'Marital_Status']:
    moda = df_preparado[col].mode()[0]
    df_preparado[col] = df_preparado[col].fillna(moda)
    print(f"- Imputada '{col}' con la moda: '{moda}'")

print(f"\nTotal de nulos restantes en el dataset: {df_preparado.isnull().sum().sum()}")
print(f"Dimensiones actuales: {df_preparado.shape[0]} filas y {df_preparado.shape[1]} columnas")
```

```python
# Cálculo exacto del porcentaje de nulos por columna (sin redondear) frente al umbral del 15%
for col in ['Education_Level', 'Income_Category', 'Marital_Status']:
    n_nulos = df_nulos[col].isnull().sum()
    pct_exacto = n_nulos / len(df_nulos) * 100
    print(f"{col}: {n_nulos} / {len(df_nulos)} = {pct_exacto:.4f}% -> supera el 15%: {pct_exacto > 15}")
```

### 8. Análisis de correlaciones (Redundancia e Irrelevancia)
Calculamos la matriz de correlación de Pearson entre las variables numéricas para ver si hay redundancia (multicolinealidad).
- Notamos que `Credit_Limit` y `Avg_Open_To_Buy` tienen una correlación de **0.996** (casi 1.0). Esto pasa porque contablemente el cupo disponible es el cupo total menos el saldo rotativo. Por tanto, eliminamos `Avg_Open_To_Buy` por redundancia.
- Otras relaciones como `Customer_Age` y `Months_on_book` (0.79) o `Total_Trans_Amt` y `Total_Trans_Ct` (0.81) son altas pero quedan por debajo del umbral de 0.85 que usamos para considerar redundancia, por eso las conservamos.


```python
# Matriz de correlación
plt.figure(figsize=(10, 8))
sns.heatmap(df_preparado.select_dtypes(include=[np.number]).corr(), annot=True, fmt=".2f", cmap="Blues")
plt.title("Matriz de Correlación entre Variables Numéricas", fontweight='bold')
plt.show()

# Eliminamos la variable redundante Avg_Open_To_Buy (corr = 0.996 con Credit_Limit)
df_final = df_preparado.drop(columns=['Avg_Open_To_Buy']).copy()
print(f"Dimensiones finales tras eliminar 'Avg_Open_To_Buy': {df_final.shape[0]} filas x {df_final.shape[1]} columnas")
```

```python
# Creamos una copia para el análisis de relevancia convirtiendo la variable objetivo a binaria
df_relevancia = df_final.copy()
df_relevancia['Attrition_Numeric'] = df_relevancia['Attrition_Flag'].apply(lambda x: 1 if x == 'Attrited Customer' else 0)

# Calculamos la correlación de las variables numéricas con la variable objetivo
correlaciones_relevancia = df_relevancia.select_dtypes(include=[np.number]).corr()['Attrition_Numeric'].drop('Attrition_Numeric')
correlaciones_relevancia = correlaciones_relevancia.sort_values(ascending=False)

# Graficamos la relevancia (correlación con Attrition_Flag)
plt.figure(figsize=(10, 6))
sns.barplot(x=correlaciones_relevancia.values, y=correlaciones_relevancia.index, palette='RdBu_r')
plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
plt.title("Relevancia de Variables Numéricas con la Deserción (Attrition_Flag)", fontsize=14, fontweight='bold')
plt.xlabel("Coeficiente de Correlación de Pearson", fontsize=12)
plt.ylabel("Variables Numéricas", fontsize=12)
plt.tight_layout()
plt.show()
```

#### Conclusión de la relevancia de las variables numéricas
La gráfica muestra que las variables más relevantes son las del comportamiento transaccional. `Total_Trans_Ct` tiene la correlación más fuerte con la deserción (-0.37), seguida de `Total_Ct_Chng_Q4_Q1` (-0.29), `Total_Revolving_Bal` (-0.26), `Avg_Utilization_Ratio` (-0.18), `Total_Trans_Amt` (-0.17), `Total_Relationship_Count` (-0.15) y `Total_Amt_Chng_Q4_Q1` (-0.13), todas negativas, lo que indica que los clientes que usan menos la tarjeta y tienen menos productos con el banco son los que más abandonan. En sentido contrario, `Contacts_Count_12_mon` (0.20) y `Months_Inactive_12_mon` (0.15) se relacionan positivamente con el abandono, porque más contactos con el banco y más meses inactivos anticipan la salida del cliente.

Las menos relevantes son `Months_on_book` (0.01), `Customer_Age` (0.02), `Dependent_count` (0.02) y `Credit_Limit` (-0.02), con correlaciones prácticamente nulas. Aun así decidimos no eliminar ninguna, porque la correlación de Pearson solo mide relaciones lineales y estas variables pueden aportar en combinación con otras o con relaciones no lineales que sí capturan los árboles y los ensambles, y porque forman parte del conjunto de predictoras registrado. `Avg_Open_To_Buy` ya se había eliminado antes por redundancia y no por irrelevancia.

A continuación se evalúa la relevancia de las variables categóricas, que no entran en la correlación de Pearson, mediante la prueba de independencia chi cuadrado y la V de Cramér como medida del tamaño de la asociación.


```python
# Relevancia de las variables categóricas frente a Attrition_Flag con chi-cuadrado y V de Cramér
cat_cols = ['Gender', 'Education_Level', 'Marital_Status', 'Income_Category', 'Card_Category']
filas_chi2 = []
for col in cat_cols:
    tabla = pd.crosstab(df_final[col], df_final['Attrition_Flag'])
    chi2, p_valor, gl, _ = stats.chi2_contingency(tabla)
    v_cramer = np.sqrt(chi2 / (tabla.values.sum() * (min(tabla.shape) - 1)))
    tasa = pd.crosstab(df_final[col], df_final['Attrition_Flag'], normalize='index')['Attrited Customer'] * 100
    filas_chi2.append({'Variable': col, 'Chi-cuadrado': round(chi2, 2), 'gl': gl,
                       'p-valor': round(p_valor, 4), 'V de Cramér': round(v_cramer, 4),
                       'Significativa (p < 0.05)': 'Sí' if p_valor < 0.05 else 'No',
                       'Tasa de abandono por categoría (%)': tasa.round(2).to_dict()})

tabla_chi2 = pd.DataFrame(filas_chi2).sort_values('V de Cramér', ascending=False).reset_index(drop=True)
tabla_chi2
```

#### Interpretación y decisión para las variables categóricas
Con un nivel de significancia de 0.05, `Gender` (p = 0.0002), `Income_Category` (p = 0.012) y `Education_Level` (p = 0.046) muestran una asociación estadísticamente significativa con la deserción, mientras que `Marital_Status` (p = 0.139) y `Card_Category` (p = 0.525) no la muestran. Concluimos que las variables categóricas aportan poca información por sí solas comparadas con las variables transaccionales. Aun así no eliminamos ninguna, porque forman parte del conjunto de 18 predictoras registrado, porque el chi cuadrado solo evalúa la relación de cada variable de forma aislada y no sus interacciones con las demás, y porque su costo dentro de los modelos es bajo. `Marital_Status` y `Card_Category` quedan identificadas como las candidatas más débiles, y su aporte real se revisará con la importancia de variables de los modelos en el notebook de modelado.
